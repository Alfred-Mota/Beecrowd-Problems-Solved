No problema temos informações valiosas:
* A distância a ser percorrida pelos ladrões é perpendicular e possui 12 milhas de distância fixos
* A distância entre os ladrões e a guarda é de D milhas

Devemos determinar se a guarda consegue pegar os ladrões se:
* A velocidade de fuga for VF
* A velocidade da guarda for VG

Da física sabemos que a distancia final S dada uma velocidade V , tempo T e distancia inicial ${S_0}$ é:
$$ {S = S_0 + V*T} $$

O tempo de percursso é dado por:

$$ T = {\frac{S - S_0}{V}} $$

A distância a ser percorrida pela guarda é de, no maximo:
$$D_G = {\sqrt(12^2 + D^2)}$$

Substituindo a distância maxima DG:
$$ T_G = {\frac{D_G}{VG}} $$

O tempo gasto pelos ladrões será:
$$ T_F = {\frac{12}{VF}} $$

Para a policia capturar os ladrões basta a seguinte condição:

$${T_G \leq T_F}$$


```python
import math

while True:
    try:
        D,VF,VG = list(map(int, (input().strip().split())))
        distancia_total = 12

    
        tf = (distancia_total)/(VF)
        Dg = math.sqrt((D*D + 144))
        tg = (Dg)/(VG)

        if(tg <= tf): 
            print("S")

        else: 
            print("N")

    except EOFError:
        break
```

```python
import math
import sys

# Usar sys.stdin facilita o tratamento de múltiplas linhas
for linha in sys.stdin:
    if not linha.strip(): # Pula linhas vazias se houver
        continue
        
    try:
        D, VF, VG = map(int, linha.split())
        
        # Distância total que os ladrões percorrem até o limite das águas
        distancia_ladrão = 12
        tf = distancia_ladrão / VF
        
        # Pitágoras para a Guarda (Catetos: D e 12)
        dg = math.sqrt(D**2 + 144)
        tg = dg / VG

        # Condição de captura
        if tg <= tf:
            print("S")
        else:
            print("N")

    except ValueError:
        # Caso haja algum problema na conversão dos dados da linha
        break
```
