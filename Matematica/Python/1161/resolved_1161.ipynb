O problema possui o seguinte desafio: __Calcular a soma de fotorial entre dois numeros__.

Sejam M e N, ambos ${0 \leq M,N \geq 20}$, calcular: M! + N!.

O fatorial é a seguinte conta: M! = M * (M-1) * (M-2) * ... * 1

Se ${M = N+X}$ -> 

* M * (M-1) * (M-2) * ...* (M-X+1) *(M-X) * ... * 1 + N * (N-1) * (N-2) * ... * 1

*  M * (M-1) * (M-2) * ...* (N+1) * N * (N-1) * (N-2) * ... * 1 + N * (N-1) * (N-2) * ... * 1

* N! * ( M * (M-1) * (M-2) * ... * (N+1) + 1)

* Dessa forma calculamos o fatorial do menor numero


```python
import math
import sys

for linha in sys.stdin:
    
    entrada = list(map(int, linha.split()))
    M,N = [max(entrada), min(entrada)]
    if(M == 0):M = 1
    if(N == 0):N = 1
    N_fatorial = math.factorial(N)
    fatorial = 1
    for i in range(N+1, M+1):
        fatorial*=i
    result = (fatorial+1)*N_fatorial
    print(result)
```
