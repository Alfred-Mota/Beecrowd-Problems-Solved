### Blobs

O problema descreve um processo de decaimento exponencial. A cada dia, a quantidade de comida $C$ é reduzida pela metade. Queremos encontrar o número de dias $N$ necessário para que a quantidade final seja menor ou igual a 1 kg.

Dados do enunciado:
* Os blobs comem metade (1/2) de seu suprimento de comida todos os dias

Dado a quantidade inicial de comida disponivel C kg, quantos dias até restar 1 kg ou menos de comida ?

1) Ao fim do primeiro dia C/2
2) Fim do segundo dia C/4
3) N-esimo dia C/${2^N}$

$${(\frac{C}{2^N}) \leq 1}$$
$${2^N \geq C}$$
$${log_2(2^N) \geq log_2(C)}$$
$${N * log_2(2) \geq log_2(C)}$$
$${N * 1 \geq log_2(C)}$$
$${N \geq log_2(C)}$$


Para resolver o problema basta determina o proximo inteiro partindo do logaritmo na base 2 de ${C}$

```python
import math
import sys

input = sys.stdin.readline

testes = int(input())
for _ in range(testes):
    C = int(input())
    C_log = math.log2(C)
    N = math.ceil(C_log)
    print(f'{N} dias')
```
