## 📘 Exercício 1047 — Tempo de Jogo com Minutos

O jogo tem duração mínima de **1 minuto** e duração máxima de **24 horas**.

---

### 📥 Entrada
Quatro números inteiros representando:
- hora inicial  
- minuto inicial  
- hora final  
- minuto final  

---

### 📤 Saída
Exiba a seguinte mensagem:

> **O JOGO DUROU XXX HORA(S) E YYY MINUTO(S)**

---

## 🧠 Solução

### Passo 1
Inicialmente, a entrada é coletada utilizando o pacote `sys`.

---

### Passo 2
Todas as horas são convertidas para **minutos**, permitindo que os cálculos sejam feitos em uma única base de referência.  
Em seguida, calcula-se a diferença entre o tempo de início e término do jogo.

---

### Passo 3
Verifica-se o resultado da diferença entre os horários, considerando os seguintes casos:

#### 🔹 Caso 1 — Horário final maior que o inicial
- A diferença entre os horários é positiva.
- O ponteiro do relógio **não cruzou a meia-noite**.
- O tempo de jogo corresponde à diferença direta entre horas e minutos.

---

#### 🔹 Caso 2 — Horário final menor ou igual ao inicial
- A diferença entre os horários é **zero ou negativa**.
- O ponteiro do relógio **cruzou a meia-noite** ou o jogo teve duração inferior a uma hora.

##### ➤ Quando `ΔH == 0`
- Pode indicar que o jogo durou **24 horas** ou **menos de uma hora**.
- Se `ΔM == 0`, então o jogo durou exatamente **24 horas**.

##### ➤ Quando `ΔH < 0`
- Indica que o ponteiro cruzou a meia-noite.
- Nesse caso, somam-se **24 horas** ao tempo total do jogo.
- Se `ΔM > 0`, o tempo final será:  
  **24 horas + ΔH + ΔM**


```python
import sys
input = sys.stdin.readline
h1,m1,h2,m2 = list(map(int, input().split()))

deltaH = (h2 - h1)*60
deltaM = m2 - m1

if(deltaH > 0): 
    endHour_min = deltaH + deltaM

elif(deltaH == 0):
    #Um dia completo
    if(deltaM == 0):
        endHour_min = 24*60 
    #23horas de Jogo
    elif(deltaM < 0):
        endHour_min = 24*60 + deltaM
    #Menos de uma hora de jogo
    else:
        endHour_min = deltaM 

else:
    #O ponteiro passou meia noite
    deltaH = deltaH + 24*60
    endHour_min = deltaH + deltaM

hour = endHour_min//60
min = (endHour_min/60 - hour)*60

print(f'O JOGO DUROU {hour} HORA(S) E {min:.0f} MINUTO(S)')
```
