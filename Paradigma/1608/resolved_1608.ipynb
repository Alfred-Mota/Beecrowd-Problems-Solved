### Bolos da Maria

Para fazer um bolo, Dona Maria precisa de certa quantidade de alguns ingredientes diferentes. Cada ingrediente tem um __custo fixo por unidade__. Ela tem uma quantia de __dinheiro D máxima para gastar__ na compra dos ingredientes. Dentre os tipos de bolos que existem, você deve escolher __apenas um tipo, de maneira a maximizar a quantia de bolos__.

Calcule o número máximo de bolos de um único tipo que podem ser confeccionados.

Entrada:
* T numeros de teste
* três números inteiros indicando o dinheiro D, o número de ingredientes I existentes e a quantidade de tipo de bolos B existentes
* A próxima linha conterá I números inteiros indicando o preço da unidade de cada ingrediente
* Seguem B linhas seguirão descrevendo cada bolo:
    * inicialmente há um número Qi que indicará quantos ingredientes diferentes serão necessários.
    * Logo em seguida teremos Qi __PARES__ de números indicando respectivamente o índice do ingrediente e a quantidade necessária, todos na mesma linha separados por espaços.

O preço total será aquele que leva em consideração a quantidade do ingrediente vezes a sua quantidade para a receita

Separando indices e quantidades

```python
import sys
import math
input = sys.stdin.readline

T = int(input())

for _ in range(T):
    #Dinheiro D, Quant Ingredientes, Quant de Bolos
    D,I,B = list(map(int, input().split(" ")))
    
    #Valor de cada ingrediente
    list_prices = list(map(int, input().split(" ")))

    #Variavel para guardar a quantidade maxima
    quantidadeMax = 0

    for j in range(B):
        #Os pares de indice e quantidade dos ingredientes necessarios
        description = list(map(int, input().split(" ")))
        
        #Separando indices
        idxs = [x for idx,x in enumerate(description) if idx&1 == 1 and idx!=0]

        #Separando quantidades
        quant = [x for idx,x in enumerate(description) if idx&1 != 1 and idx!=0]
        
        #preço inicial
        price = 0
        for i in range(len(idxs)):
            idx = idxs[i]
            #Somando os preços de cada ingrediente
            price += list_prices[idx]*quant[i]          

        #comparando os preços e arredondando para baixo
        quantidadeMax = math.floor(max(quantidadeMax, D/price))
    print(quantidadeMax)
            
```

Forma mais direta

```python
import sys
import math
input = sys.stdin.readline

T = int(input())

for _ in range(T):
    #Dinheiro D, Quant Ingredientes, Quant de Bolos
    D,I,B = list(map(int, input().split(" ")))
    
    #Valor de cada ingrediente
    list_prices = list(map(int, input().split(" ")))

    #Variavel para guardar a quantidade maxima
    quantidadeMax = 0

    for j in range(B):
        #Os pares de indice e quantidade dos ingredientes necessarios
        description = list(map(int, input().split(" ")))
        
        #preço inicial
        price = 0
        
        for i in range(1,len(description),2):
            idx = description[i]
            price += list_prices[idx]*description[i + 1]

        quantidadeMax = math.floor(max(quantidadeMax, D/price))
    print(quantidadeMax)
            
```
